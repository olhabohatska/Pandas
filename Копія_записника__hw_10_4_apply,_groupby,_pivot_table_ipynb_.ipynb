1. Дослідимо, який філіал супермаркету ('Branch') є найприбутковішим. Для цього знайдіть сумарний прибуток за кожним філіалом і виявіть, який філіал має найвищий.

```python
branch_profit=df.groupby('Branch')['gross income'].sum()
print('Сумарний прибуток за філіалами:')
print(branch_profit)
```

```python
best_branch=branch_profit.idxmax()
print(f'\nНайприбутковіший філіал:{best_branch}(прибуток:{branch_profit.max():2f})')
```

2. В якому місті знайходиться філіал з найвищим прибутком? Може в тому місці нам розмітисти ще один магазин.  
Знайдіть відповідь за допомогою функціоналу Pandas.

```python
city_profit=df.groupby('City')['gross income'].sum()
print('Сумарний прибуток за містами:')
print(city_profit)
```

```python
best_city=city_profit.idxmax()
print(f'\nМісто з найвищим прибутком: {best_city}')
```

3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoices_by_category=df.pivot_table(
    index='Branch',
    columns='Product line',
    values='Invoice ID',
    aggfunc='count'
)

invoices_by_category
```

 Питання можуть бути будь-якої складності, але їх має бути 3. Кожне питання оцінюється як 1 завдання. Без виконання цього завдання ДЗ не приймається. Якщо є питання щодо виконання - пишіть у чат 🙌


```python
#Гіпотеза 1. В який час доби клієнти роблять найбільше покупок?
df['Hour']=pd.to_datetime(df['Time']).dt.hour
hourly_sales=df.groupby('Hour')['Invoice ID'].count()
best_hour=hourly_sales.idxmax()
print(f"Пікова година покупок: {best_hour}:00")
```

Бізнес рішення: Оптимізуваи графік зміни касирів та викладку товару в пікові години, а також планувати технічні перерви у години найнижчої активності

```python
#Гіпотеза 2: Чи відрізняється середній чек та вибір способу оплати (Cash, Credit card, Ewallet) залежно від типу клієнта(Member vs Normal)?
payment_by_customer=df.pivot_table(index='Customer type', columns='Payment', values='Total',aggfunc='mean')
print(payment_by_customer)
```

Бізнес-рішення: Якщо лояльні клієнти частіше користуються електонними гаманцями або витрачають більше, можна впровадити спеціальні бонуси та кешбек за цей спосіб оплати для стимулювання повторних покупок.

```python
# Гіпотеза 3: Яка категорія товарів приносить найвищий середній рейтинг задоволеності клієнтів ?
rating_by_category=df.groupby('Product line') ['Rating'].mean().sort_values(ascending=False)
print(rating_by_category)
```

Бізнес-рішення:Категорії з найнижчим середнім рейтингом потребують перевірки якості товарів або асортименту, а категорії з високим рейтингом можна активніше рекламувати як флагманські.
